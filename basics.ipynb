Column Filtering

```python
df_trimmed = df.select("title","imdb_rating","industry")
```

```python
df_trimmed.show(3,truncate=False)
```

Row Filtering

```python
df_2000_2010 = df.filter(df['release_year'].between(2000, 2010))
```

```python
display(df_2000_2010)
```

```python
df_marvel = df.filter(df['studio'] == 'Marvel Studios')
display(df_marvel)
```

```python
unique_industry = df.select("industry").distinct()
display(unique_industry)
```

```python
unique_languages = df.select('language').distinct()
display(unique_languages)
```

```python
from pyspark.sql.functions import col

df_with_profit = df.withColumn('profit', col('revenue') - col("budget"))
display(df_with_profit)
```

```python
df.withColumnRenamed("revenue","total_revenue")
df.printSchema()
```
